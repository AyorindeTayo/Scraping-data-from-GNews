```python
from GoogleNews import GoogleNews
import pandas as pd

def fetch_ai_news_headlines():
    # Initialize the GoogleNews client
    google_news = GoogleNews()

    # Set the search query to "Artificial Intelligence"
    google_news.get_news("Artificial Intelligence") # Changed from search to get_news

    # Fetch the latest news articles (default is 10 articles)
    articles = google_news.results()  # Changed from get_news to results

    # Extract headlines and URLs
    headlines = []
    for article in articles:
        headlines.append({
            "Title": article["title"],
            "URL": article["link"] # Changed from 'url' to 'link'
        })

    # Convert the list of headlines to a DataFrame
    df = pd.DataFrame(headlines)

    # Save the DataFrame to a CSV file
    df.to_csv("ai_news_headlines.csv", index=False)
    print("News headlines saved to 'ai_news_headlines.csv'")

    return df

if __name__ == "__main__":
    # Fetch and save news headlines
    headlines_df = fetch_ai_news_headlines()

    # Display the DataFrame in Colab
    display(headlines_df)
```
